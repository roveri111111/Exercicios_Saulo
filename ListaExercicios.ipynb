### Exercício 1
Crie uma variável `idade` e atribua um valor inteiro a ela. Usando uma estrutura `if` e um operador booleano, imprima 'Maior de idade' se a `idade` for 18 ou mais, caso contrário, imprima 'Menor de idade'.

```python
idade = int(input("Digite sua idade: "))

if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")
```

### Exercício 2
Crie duas variáveis booleanas, `chuva` e `sol`. Usando um `if` e o operador `and`, imprima 'Céu nublado' se estiver chovendo E fazendo sol. Caso contrário, imprima 'Condição climática usual'.

```python
chuva = input("Esta chovendo? ").lower()
sol = input("Esta fazendo sol? ").lower()

if chuva == "sim" and sol == "sim":
    print("Ceu nublado")
else:
    print("Condiçao climatica usual")
```

### Exercício 3

### Exercício 6
Crie variáveis `temperatura` (numérica) e `umidade` (numérica). Se a `temperatura` for maior que 25 E a `umidade` for maior que 70, imprima 'Clima abafado'. Se `temperatura` for menor que 10 OU `umidade` for menor que 30, imprima 'Clima extremo'. Caso contrário, imprima 'Clima agradável'.

```python
temperatura = float(input("Digite a temperatura: "))
umidade = float(input("Digite a umidade: "))

if temperatura > 25 and umidade > 70:
    print("Clima abafado")
elif temperatura < 10 or umidade < 30:
    print("Clima extremo")
else:
    print("Clima agradavel")
```
